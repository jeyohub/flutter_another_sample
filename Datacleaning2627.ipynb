### Data Cleaning for `df_students`
Following the steps from the activity images, we will:
1. Check for missing values and duplicates.
2. Calculate the IQR (Interquartile Range) to identify outliers in quantitative columns like `exam_score` or `study_hours`.
3. Address and clean the anomalies.

```python
import pandas as pd
import numpy as np

# Mock dataset representing students_raw.csv structure seen in the activity
data = {
    'student_id': ['S001', 'S002', 'S003', 'S004', 'S005', 'S006', 'S007', 'S008'],
    'full_name': ['Ana Cruz', 'Ben Santos', 'Carla Reyes', 'Dan Lim', 'Ella Tan', 'Fred Cruz', 'Gina Alis', 'Hugo Low'],
    'age': [19, 20, np.nan, 'twenty', 21, 22, 19, 23],
    'course': ['bsit', 'BSCS', 'B.S.I.T.', 'BSIS', 'bsba', 'BSCS', 'BSIT', 'BSBA'],
    'city': ['digos', 'Davao', 'DIGOS', 'Tagum', 'Digos', 'Davao', 'Digos', 'Tagum'],
    'attendance_pct': [95.0, 90.0, 88.0, 85.0, 92.0, 93.0, 10.0, 91.0],
    'exam_score': [88.0, 85.0, 79.0, 75.0, 91.0, 2.0, 84.0, 150.0],  # 2.0 and 150.0 are potential outliers
    'study_hours': [8, 7, 6, 5, 9, 1, 6, 24],
    'enrollment_date': ['2026-08-01', '2026-08-02', '03-08-2026', '2026-08-04', '2026-08-05', '2026-08-06', '2026-08-07', '2026-08-08'],
    'email': ['ana@example.com', 'ben@example.com', 'carla@example.com', 'danexample.com', 'ella@example.com', 'fred@example.com', 'gina@example.com', 'hugo@example.com']
}

df_students = pd.DataFrame(data)
print("Missing values per column:")
display(df_students.isna().sum())
```

```python
# 2. Identify outliers using the IQR method on 'exam_score'
# Ensuring exam_score is numeric
df_students['exam_score'] = pd.to_numeric(df_students['exam_score'], errors='coerce')

q1_score = df_students['exam_score'].quantile(0.25)
q3_score = df_students['exam_score'].quantile(0.75)
iqr_score = q3_score - q1_score

lower_bound = q1_score - (1.5 * iqr_score)
upper_bound = q3_score + (1.5 * iqr_score)

outliers_students = df_students[(df_students['exam_score'] < lower_bound) | (df_students['exam_score'] > upper_bound)]

print(f"Q1 (25th Percentile): {q1_score}")
print(f"Q3 (75th Percentile): {q3_score}")
print(f"IQR: {iqr_score}")
print(f"Lower Bound (Q1 - 1.5*IQR): {lower_bound}")
print(f"Upper Bound (Q3 + 1.5*IQR): {upper_bound}")
print("\nOutliers identified via IQR:")
display(outliers_students)
```

```python
# 3. Handle anomalies and complete data cleaning for df_students
import pandas as pd
import numpy as np

# Standardize text formatting and handle structural anomalies
df_students['course'] = df_students['course'].str.upper().str.replace('.', '', regex=False)
df_students['city'] = df_students['city'].str.capitalize()

# Map written age strings to numbers
age_mapping = {'twenty': 20}
df_students['age'] = df_students['age'].replace(age_mapping)
df_students['age'] = pd.to_numeric(df_students['age'], errors='coerce')

# Fill missing values in Age using the column's median value
median_age = df_students['age'].median()
df_students['age'] = df_students['age'].fillna(median_age)

# Filter out outliers based on realistic thresholds (e.g. Exam score capped at 100 max)
df_students = df_students[(df_students['exam_score'] >= 0) & (df_students['exam_score'] <= 100)]

print("Cleaned Student DataFrame:")
display(df_students)
```
